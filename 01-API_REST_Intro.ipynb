# **Introduction**

## What is an API?

API stands for "Application Programming Interface". It is an **interface that allows a program to use the functionality of another software component** without needing to know how it works internally.

An API is therefore not necessarily used to allow two computers to communicate over a network. It can be used between different software components on the same computer.

APIs in the broad sense therefore include function libraries, classes, and modules that allow developers to use the functionality of software without having to understand its source code.

Some examples of APIs:

* The **Pandas** API allows Python programs to use functions, classes, and methods to manipulate data.

* The **Google** authentication API allows a website to let its users sign in with their Google account without the website needing to know their password.

In the first case, the API is a **local** interface between a Python program and a library. In the second case, it is a **Web API** that allows software to communicate over a network.

## What is a Web API?

In the context of the Web, the term API often refers to a **Web API**, that is, an interface that allows software to exchange data and request processing over a network.

A Web API generally allows a client program to send requests to a server and receive responses.

For example:

```text

Python program

       │

       │ HTTP request

       ▼

    Web API

       │

       │ HTTP response

       ▼

Python program

```

Web APIs generally use the **HTTP (HyperText Transfer Protocol)** protocol, created in the early 1990s notably by Tim Berners-Lee as part of the development of the World Wide Web.

## What is a REST API?

A **REST API** (REST = "Representational State Transfer") is therefore a Web API designed according to the principles of the REST architecture.

REST was notably formalized by Roy Fielding in his thesis published in 2000. [Learn more](https://en.wikipedia.org/wiki/Representational_state_transfer)

REST APIs generally use HTTP to allow a client and a server to exchange representations of resources.

> **Attention: API, REST, and HTTP refer to three different concepts.**
>
> * **API** refers to an interface that allows software to use the functionality of another software component.
>
> * **HTTP** is a communication protocol used notably on the Web.
>
> * **REST** is an architectural style that can be used to design Web APIs.

A Web API can therefore be RESTful, but not all Web APIs are necessarily RESTful.

They generally share the following characteristics:

### An API URL

A Web API is generally accessible through a **URL** (*Uniform Resource Locator*), which identifies the server and the path used to access the API's different resources.

It looks like a standard website URL.

Ex:

```
https://catfact.ninja/
```

In this URL:

* `https://` is the protocol being used;

* `catfact.ninja` is the server's domain.

### Endpoint

An **endpoint** is a URL to which a client can send a request in order to access a resource or functionality of an API.

For example:

```
https://catfact.ninja/fact
```

The *path* `/fact` provides access to a resource that returns a random fact about cats.

We can therefore distinguish:

```text

https://catfact.ninja/fact

│       │             │

│       │             └── path

│       └── domain

└── protocol

```

### HTTP Methods / Actions

An HTTP request uses an **HTTP method** that indicates the operation the client wants to perform.

Among the HTTP methods most commonly used in Web APIs are:

* The **GET** request is generally used to retrieve data from the server.

* The **POST** request is generally used to send data to the server for processing. It can notably be used to create a new resource.

* **PUT** is used to create a resource or replace its contents at a given URL.

* **DELETE** is used to request the deletion of a resource.

These methods are often associated with CRUD operations:

| HTTP Method | Common CRUD Operation |
| ----------- | --------------------- |
| `GET`       | Read                  |
| `POST`      | Create                |
| `PUT`       | Update / Replace      |
| `DELETE`    | Delete                |

> This mapping is a pedagogical simplification. HTTP methods have a more precise semantics than these simple CRUD operations.

When you enter a URL in a browser (Chrome, Firefox, Safari...), it generally performs an HTTP **GET** request to retrieve the corresponding resource.

### Response

When an API receives a request, it returns an **HTTP response**. Example:

```
HTTP/1.1 200 OK

Content-Type: application/json

{
    "fact": "Owning a cat is actually proven to be beneficial for your health.",
    "length": 65
}
```

The response notably contains a **status code** indicating the result of the request and may also contain data.

#### HTTP Status Codes

HTTP status codes are three-digit numbers that indicate the result of an HTTP request. The most common ones are:

| Code | Meaning      |
| ---- | ------------ |
| 200  | OK           |
| 201  | Created      |
| 400  | Bad Request  |
| 401  | Unauthorized |
| 403  | Forbidden    |
| 404  | Not Found    |

The data returned by a Web API is often in **JSON** format (*JavaScript Object Notation*).

JSON is a text format widely used to exchange data between applications. Its syntax resembles that of Python dictionaries, but JSON and Python dictionaries are two different things.

### Query Parameters

After specifying the *path* of the endpoint, it is possible to pass **query parameters**.

Query parameters are added to the URL after a `?`.

The syntax is:

```text
?name=value
```

For example, we can limit the length of the random cat fact by using the `max_length` parameter:

https://catfact.ninja/fact?max_length=48

In this URL:

* `/fact` is the *path*;

* `max_length=48` is a *query parameter*;

* `max_length` is the parameter name;

* `48` is the parameter value.

If you press F5 to refresh your browser, you can see that no response contains more than 48 characters.

If you want to pass multiple parameters, we use `&` as a separator.

For example:

```text
https://example.com/search?query=python&limit=10
```

Here, we have two parameters:

```text
query=python

limit=10
```

### Payload

The term **payload** generally refers to the **useful data carried in the body of an HTTP request or response**.

It is important to distinguish the payload from query parameters.

For example, in:

```text
https://catfact.ninja/fact?max_length=48
```

`max_length=48` is a **query parameter**, not a payload.

A GET request can therefore contain parameters in its URL without necessarily having a payload in its body.

On the other hand, a POST request can, for example, send data in its body:

```http
POST /users

Content-Type: application/json

{
    "name": "Alice",
    "age": 25
}
```

In this example, the JSON data contained in the *body* constitutes the request payload.

Similarly, the API response can contain a payload:

```json
{
    "name": "Alice",
    "age": 25
}
```

The term *payload* is nevertheless used somewhat loosely depending on the context. It generally refers to the useful data carried by a request or response, as opposed to information required for its transport or processing, such as certain HTTP headers.

## Python

The `requests` library allows a Python program to send HTTP requests and process their responses.

It is particularly convenient for using Web APIs.

For example:

```python
import requests

response = requests.get("https://catfact.ninja/fact")

print(response.status_code)

print(response.json())
```

Here:

* `requests.get()` sends an HTTP `GET` request;

* `response` contains the server's response;

* `response.status_code` provides access to the HTTP status code;

* `response.json()` converts the JSON content of the response into Python objects.


````python
import requests

url = 'https://catfact.ninja/'
fact_endpoint = 'fact'
response = requests.get(url + fact_endpoint) # Let's use the 'get' method
````

````python
type(response)
````

Un objet "Response" possède de nombreuses propriétés et méthodes différentes. Nous allons utiliser ``.json()`` qui convertira la réponse en dictionnaire.

A Response object has many different properties and methods. We are going to use the ``.json()`` which will convert the response to a dictionnary.

````python
r = response.json()
````

````python
print(r)
````

````python
type(r)
````

````python
r['fact']
````

````python
r['length']
````

## Parameters

If we want to use parameters we can just add them at the end of the ``requests.get()`` arguments, like this :

````python
response = requests.get('https://catfact.ninja/fact?max_length=48')
print(response.json())
````

But sometimes it's best to give a dict of arguments, the syntax is then :

````python
my_params = {'max_length' : 24}

response = requests.get('https://catfact.ninja/fact', params=my_params)
print(response.url)
print(response.json())
````
